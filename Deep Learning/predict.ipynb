```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open('label_encoder.pkl', 'rb') as file:
    label = pickle.load(file)

with open('onehot_encoder.pkl', 'rb') as file:
    onehot = pickle.load(file)

with open('scalar_encoder.pkl', 'rb') as file:
    scalar = pickle.load(file)
```

```python
model = load_model('model.h5')
```

```python
user = {'Pclass': 3, 'Sex': 'female', 'SibSp': 1, 'Parch': 1, 'Fare': 11, 'Embarked': 'Queenstown'}
```

```python
user = pd.DataFrame([user])
```

```python
user['Sex'] = label.transform(user['Sex'])
```

```python
embarked = onehot.transform(user[['Embarked']])
```

```python
embarked = pd.DataFrame(embarked, columns = onehot.get_feature_names_out())
```

```python
user = pd.concat([user.drop(columns = ['Embarked']),embarked], axis=1)
```

```python
user[['Pclass', 'SibSp', 'Parch', 'Fare']] = scalar.transform(user[['Pclass', 'SibSp', 'Parch', 'Fare']])
```

```python
y = model.predict(user)
```

```python
y
```

```python
if y>0.5:
    print('The passenger is likely to survive.')
else:
    print('The passenger is likely to not survive.')
```
