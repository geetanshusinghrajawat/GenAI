# Training a Word2Vec Model on Our Personal Data

```python
# File Handling

with open('Book1.txt', 'rt',encoding='utf-8') as b1:
    h1 = b1.read()
with open('Book2.txt', 'rt',encoding='utf-8') as b2:
    h2 = b2.read()
with open('Book3.txt', 'rt',encoding='utf-8') as b3:
    h3 = b3.read()
with open('Book4.txt', 'rt',encoding='utf-8') as b4:
    h4 = b4.read()
with open('Book5.txt', 'rt',encoding='utf-8') as b5:
    h5 = b5.read()
with open('Book6.txt', 'rt',encoding='utf-8') as b6:
    h6 = b6.read()
with open('Book7.txt', 'rt',encoding='utf-8') as b7:
    h7 = b7.read()
    
```

```python
data = h1+h2+h3+h4+h5+h6+h7
```

```python
len(data.split()) #Total number of words our harry potter data corpus
```

## Applying NLP Preprocessing
#### ---> 1. Tokenization-> Sentence Tokenization
#### ---> 2. Stopwords Removal
#### ---> 3. LowerCasing
#### ---> 4. Rest of Basic Pre-Processing Steps

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
from nltk.corpus import stopwords
import gensim
from sklearn.decomposition import PCA
import plotly.io as pio
import plotly.express as px
```

```python
potter = []
data = sent_tokenize(data) # Dividing the data into sentences
```

```python
len(data)
```

```python
for i in data:
    potter.append(simple_preprocess(i))
```

```python
potter
```

```python
# Stopwords helps in removing the commonly used words in the language while keeping the meaning of the sentence intact.
# It also, helps in reducing the size of the data and compuation time.
# Saves memory.

stopwords = set(stopwords.words('english'))
```

```python
def stopword_removal(text):
    cleaned_text = []
    for sentence in text:
        new_sentence = []
        for word in sentence:
            if word.lower() not in stopwords:
                new_sentence.append(word)
        cleaned_text.append(new_sentence)
    return cleaned_text
```

```python
cleaned_potter = stopword_removal(potter)
```

## Applying Word2Vec on the MODEL

```python
model = gensim.models.Word2Vec(window = 10,min_count = 2)
```

```python
model.build_vocab(cleaned_potter)
```

```python
model.train(cleaned_potter, total_examples = model.corpus_count, epochs = model.epochs)
# total_examples is the total number of sentences present in the data (cleaned_potter)
# epochs is how many times the model is going to iterate over the data
```
